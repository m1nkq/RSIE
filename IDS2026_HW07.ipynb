The second step of CRISP-DM is data understanding. The goal is to collect, describe, explore, and verify the quality of the data to ensure it’s suitable for the analytical or modeling tasks ahead.

**a. Describe the data. Number of rows and columns, what does one row represent.**

```python
import pandas as pd

# Load the dataset
data = pd.read_csv("lo_2011_2026.csv",
                   sep=";", dtype={"Juhtumi nr": "string"})
# Convert dates
data["Toimumisaeg"] = pd.to_datetime(data["Toimumisaeg"])

# Keep only accidents from 2013 to 2025
data = data[data["Toimumisaeg"].dt.year.between(2013, 2025)]

# Check and remove duplicates
print("Duplicates:", data.duplicated().sum())
data = data.drop_duplicates()

# Show basic information
print("Rows and columns:", data.shape)
print("Missing values:", data.isnull().sum().sum())
display(data.head())
```

**Answer:** The original dataset contains 23,869 rows and 54 columns. We selected the period 2013–2025, excluding 2011–2012 because their geographical data was not corrected in the same way as later years, and 2026 because it is incomplete.

One duplicate was found and removed, leaving 19,867 accidents and 54 columns. The filtered dataset contains 89,822 missing values across all columns.

The dataset only includes accidents involving injuries or fatalities. Therefore, we can compare accident characteristics and proportions, but we cannot determine the overall risk of accidents without traffic volume data.

**c. First shots at visualizing the data**

Take a look at the data as plots to get more insight to the patterns, trends, and relationships. This could be histograms, scatter plots, box plots, ect. What did you gather? What stands out?

```python
import matplotlib.pyplot as plt

# Create columns for analysis
data["Year"] = data["Toimumisaeg"].dt.year
data["Fatal accident"] = data["Hukkunuid"] > 0

# 1. Accidents by year
data["Year"].value_counts().sort_index().plot(kind="bar")
plt.title("Road accidents by year (2013–2025)")
plt.xlabel("Year")
plt.ylabel("Number of accidents")
plt.show()

display(data["Year"].value_counts().sort_index())

# 2. Fatal accidents by lighting
lighting = data.groupby("Valgustus")["Fatal accident"].mean() * 100
lighting.plot(kind="bar", color="orange")
plt.title("Fatal accidents by lighting conditions")
plt.ylabel("Fatal accidents (%)")
plt.show()

display(lighting.round(2))

# 3. Fatal accidents by weather
weather = data.groupby("Ilmastik")["Fatal accident"].agg(["mean", "count"])
weather = weather[weather["count"] >= 100]
weather = weather["mean"] * 100
weather.plot(kind="bar", color="green")
plt.title("Fatal accidents by weather conditions")
plt.ylabel("Fatal accidents (%)")
plt.show()

display(weather.round(2))
```

**Answer:** We created three bar charts to explore road accident patterns between 2013 and 2025.

The first graph shows that the number of recorded accidents generally increased after 2020, reaching its highest point in 2024 with 1,920 accidents. In 2025, the number decreased to 1,616.

The second graph shows that approximately 5.7% of accidents occurring in darkness were fatal, compared with around 3.3% in daylight.

The third graph shows differences in fatal accident percentages across weather conditions. Snowy conditions had the highest proportion (about 7.5%), followed by foggy conditions (6.9%) and rainfall (5.8%). Only weather categories with at least 100 accidents were included.

Overall, these visualizations reveal differences in accident characteristics and fatality proportions. However, they do not prove that specific weather or lighting conditions are more dangerous, because traffic volume data is unavailable.
